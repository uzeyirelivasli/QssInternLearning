```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(1,10,100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.sin(x) * np.cos(x)

plt.plot(x,y1, label='sin')
plt.plot(x,y2, label='cos')
plt.plot(x,y3, label='sin*cos')
plt.legend()
```

```python
categories=['A','B','C','D']
values=[12,25,20,35]

plt.bar(categories,values)
```

## Pie chart

```python
categories = ['North','South','East','West']
sizes = [30,25,20,25]
explode = (0.1,0,0,0) #explode first one

plt.pie(sizes,labels=categories,explode=explode)
```
