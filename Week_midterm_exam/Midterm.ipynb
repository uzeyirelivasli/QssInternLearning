```python
import pandas as pd
```

```python
df = pd.read_excel('tips.xlsx')
df.head()

df.shape
#244 rows 9 columns
```

```python
# 1. Call our dataset and get its first 8 and last 4 rows, show number of observations
df.head(8)
df.tail(4)
```

```python
# Report number of null values for each column. (3 points)
# a. Remove columns with null values. 

# df.isnull().sum()
df = df.dropna()
```

```python
df.isnull().sum()
```

```python
# What is the most frequent day of given tips? 
df.groupby('day')['tip'].count().sort_values(ascending=False).head(1)
```

```python
# Sort values by total bill and tip at restaurant in descending order. 
df_sorted = df.sort_values(['total_bill', 'tip'], ascending=False)
df_sorted.head()
```

```python
# Group by data by time and sex and find mean. (5 points)
# a. Which customers pay more than others?

result = df.groupby(['time', 'sex'])['tip'].mean()
```

```python
# Filter total bill greater than 20 and tip less than 3
new_frame= df[(df['total_bill']>20) & (df['tip'] < 3)]
new_frame.sample(10)
```

```python
# Filter male customers with size equal to 2. 
male_filter = df[df['size'] == 2]
male_filter.sample(10)
```

```python
# Find most frequent gender using “if” function.
male_count = (df['sex'] == 'Male').sum()
female_count = (df['sex'] == 'Female').sum()

if male_count > female_count:
    print('Male')
elif female_count > male_count:
    print('Female')
else:
    print('Tie')
```

```python
# Find number of bills greater than daily average. 
avg_bill = df['total_bill'].mean() 
# Count bills greater than average
count_above_average = (df['total_bill'] > avg_bill).sum()
count_above_average
```
