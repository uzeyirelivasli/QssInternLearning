```python
df.head()       # ilk 5 satır
df.tail()       # son 5 satır
df.info()       # sütun tipleri, null sayısı
df.describe()   # istatistiksel özet
df.shape        # (satır, sütun)
df.columns      # sütun isimleri
df.dtypes       # veri tipleri
```

```python
df['yas']                    # tek sütun
df[['isim', 'yas']]          # birden fazla sütun
df.loc[0]                    # label ile satır
df.iloc[1]                   # index numarasıyla satır
df[df['yas'] > 25]           # koşullu filtreleme
df.query('yas > 25')         # aynı işlem, farklı sözdizim
```

```python
# Eksik Veri
df.isnull().sum()            # her sütunda kaç NaN var
df.dropna()                  # NaN olan satırları sil
df.fillna(0)                 # NaN'ları doldur
```

```python
df.groupby('companies')[['sales']].sum()
```

```python
df.groupby('companies')['sales'].sum()      # total sales per company
df.groupby('companies')['sales'].mean()     # average sales per company
df.groupby('companies')['sales'].count()    # number of records per company
df.groupby('companies')['sales'].agg(['sum', 'mean', 'count'])  # multiple at once
```
