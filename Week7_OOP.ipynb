## Introduction OOP

```python
class Vehicle:
    color = "white"

    def __init__(self, maxspeed,capacity):
        self.maxspeed = maxspeed
        self.capacity= capacity

    def describe(self):
        return f"A {self.color} vehicle with max speed {self.maxspeed}"

    def default_fare(self):
        return self.capacity * 35
        
        


class Car(Vehicle):
    def __init__(self, maxspeed,capacity,doors):
        super().__init__(maxspeed,capacity)
        self.doors = doors

    def describe(self):
        return f"A {self.color} car with max speed {self.maxspeed} and {self.doors} doors"

        
class Train(Vehicle):

    def default_fare(self):
        return self.capacity * 35 * 1.1
    
    

car2 = Train(200, 4)
print(car2.default_fare())
```
