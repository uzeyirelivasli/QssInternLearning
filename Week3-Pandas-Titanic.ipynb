```python
df.head()
```

```python
# Find the survival rate (%) by gender — group by Sex, mean of Survived.
df.groupby('Sex')['Survived'].mean()
```

```python
# Find the survival rate by Pclass.
df.groupby('Pclass')['Survived'].mean()
```

```python
# Fill missing Age values with the median age.
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
# Drop the Cabin column (too many missing values) — or count what % is missing first.
null_cabin_percentage = df['Cabin'].isnull().sum() / len(df['Cabin']) * 100
null_cabin_percentage

# drop the cabin column
df.drop(columns='Cabin', inplace = True)
df
```

```python
# Create a new column FamilySize = SibSp + Parch + 1.
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1
df.head()
```

```python
# Create a new column IsAlone — 1 if FamilySize == 1, else 0.
df['IsAlone'] = np.where(df['FamilySize'] ==1, 1, 0)
df.head()
```

```python
# Extract passenger titles (Mr, Mrs, Miss, etc.) from the Name column using string extraction.

df['Title'] = df['Name'].str.extract(r',\s*(\w+)\.')
df.head()
```

```python
# Find the average Fare paid per Pclass.
df.groupby('Pclass')['Fare'].agg(['mean', 'min', 'max', 'count'])
```

```python
# Find the oldest and youngest survivor (full row, not just age).
survivors = df[df['Survived'] == 1] 

oldest_survivor = survivors.loc[survivors['Age'].idxmax()]
youngest_survivor = survivors.loc[survivors['Age'].idxmin()]
```

```python
# Group by both Sex and Pclass, and find the survival rate for each combination.

df.groupby(['Sex','Pclass'])['Survived'].mean()
```

```python
# Find how many passengers embarked from each port (Embarked), and which port had the highest survival rate.
embarked_group = df.groupby('Embarked')['PassengerId'].size()
embarked_surv_rate = df.groupby('Embarked')['Survived'].mean().idxmax()
embarked_surv_rate
```

```python
# Bucket Age into groups (e.g. Child: 0–12, Teen: 13–19, Adult: 20–59, Senior: 60+) using pd.cut, then find survival rate per age group.
age_bins = [0,12,19,59,100]
labels = ['Child','Teen','Adult','Senior']
# Apply pd.cut
categories = pd.cut(df['Age'], bins=age_bins, labels=labels)
df.groupby(categories)['Survived'].mean()
```
