```python
# Find how many Movies vs TV Shows are in the dataset, and what percentage each makes up.
# drop one broken data
df = df[df['type'] != '1944']
(df.groupby('type').size() / len(df) * 100).round(2)
```

```python
# Find the average duration_minutes for Movies only (remember: TV Shows have NaN here, so you'll need to filter first).

# duration minutes was kept as strings, converted to numeric
df['duration_minutes'] = pd.to_numeric(df['duration_minutes'], errors='coerce')

movie_df = df[df['type']=='Movie']
movie_df['duration_minutes'].mean()
```

```python
# Convert date_added to a proper datetime column, then extract just the year and month into two new columns.

# I will drop release_year because I will add Year and month
df['date_added'] = pd.to_datetime(df['date_added'])
df['year'] = df['date_added'].dt.year
df['month'] = df['date_added'].dt.month
```

```python
df.head(10)
```

```python
# Find how many titles were added to Netflix per year — which year had the most additions?
df['year_added'] = pd.to_datetime(df['date_added'], errors='coerce').dt.year
df.groupby('year_added')['title'].count().sort_values(ascending=False)
```

```python
# Find the gap between release_year and the year it was date_added — create a new column years_to_netflix showing
# how many years after release the title was added. Which title had the longest gap?

df['years_to_netflix'] = df['year_added'] - df['release_year']
# df['years_to_netflix'].sort_values(ascending=False)

# which movie is it
idx = df['years_to_netflix'].idxmax()

# you can add comma and type the columns you want
df.loc[idx]
```

```python
# Group by rating, and find the average duration_minutes per rating — do R-rated movies tend to run longer than TV-Y7 ones?
df['duration_minutes'] = pd.to_numeric(df['duration_minutes'], errors='coerce')

# searching in Movie dataset due to tv shows using duration_seasons instead of duration_minutes
movie_df = df[df['type']=='Movie']

rating_duration = movie_df.groupby('rating')['duration_minutes'].mean()
rating_duration['R'] > rating_duration['TV-Y7']
```

```python
# Find the most common rating overall, and separately, the most common rating for Movies vs for TV Shows.

# creating df instance for series
series_df = df[df['type']=='TV Show']

df.groupby('rating')['title'].count().idxmax()

movie_df.groupby('rating')['title'].count().idxmax()
series_df.groupby('rating')['title'].count().idxmax()   # TV shows only

```

```python
# Create a new column decade from release_year (e.g. 2019 → "2010s") using either pd.cut or 
# a manual function, then find how many titles come from each decade.

bins = list(range(1920, 2031, 10))

# -1 because always one fewer range than there are edges.
labels = [f'{b}s' for b in bins[:-1]]
df['decade'] = pd.cut(df['release_year'], bins=bins, labels=labels, right=False)
df.groupby('decade')['title'].count().sort_values(ascending=False)
```

```python
# Find all titles containing the word "Love" in the title column (string search), and show just title, type, and release_year for those matches.
title_love = df[df['title'].str.contains('love', case=False, na=False)]
title_love[['title', 'type', 'release_year']]
```

```python
# Sort the dataset by duration_minutes descending and find the 5 longest movies — 
# then do the same for duration_seasons to find the TV shows with the most seasons.

df['duration_seasons'] = pd.to_numeric(df['duration_seasons'], errors='coerce')
# filter out a tv show garbage season data
df = df[df['duration_seasons'] < 100]

df.sort_values('duration_minutes', ascending=False)[['title', 'duration_minutes']].head()
df.sort_values('duration_seasons', ascending=False)[['title', 'duration_seasons']].head()
```
