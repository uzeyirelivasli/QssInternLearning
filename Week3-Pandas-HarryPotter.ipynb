```python
import pandas as pd
import requests

base = 'https://potterapi-fedeperin.vercel.app/en'
# Books
books = requests.get(f'{base}/books').json()
df_books = pd.DataFrame(books)

# Titles uppercase
df_books.columns = df_books.columns.str.upper()

# drop number & index column
df_books.drop(columns='NUMBER', inplace=True)
df_books.drop(columns='INDEX', inplace=True)


df_books['RELEASEDATE'] = pd.to_datetime(df_books['RELEASEDATE'])
# for dt.year releasedate should be datetime format first
df_books['year'] = df_books['RELEASEDATE'].dt.year

# book with the most pages
# df_books.loc[df_books['PAGES'].idxmax()]

# ordering for pages
# df_books = df_books.sort_values('PAGES', ascending=False)

df_books
```
